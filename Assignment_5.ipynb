```python
## 1>> Check whether a number is even or odd ?
num=eval(input("Enter a number :"))
if num%2==0:
    print('even')
else:
    print("odd")
    
```

```python
#2>> Check whether a number is positive, negative, or zero.
num=eval(input("Enter a number :"))
if num>0:
    print("positive")
elif num<0:
    print("negative")
else:
    print("zero")
```

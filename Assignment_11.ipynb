```python
# Q8) str1='hello how are you, hello im good'
#     take any sentence like above find the most reptaed word

str1='hello how are you, hello im good'

words=str1.split()

max_count=0
max_word=''

for i in words:
    count=0

    for j in words:
        if i ==j:
            count =count+1


        if count > max_count:
            max_count=count
            max_word = i
print(max_count)
print(max_word)
```

```python
def hillnum(list1):
    decreasing = False

    for i in range(len(list1) - 1):

        if list1[i] < list1[i+1]:

            if decreasing:
                return False

        elif list1[i] > list1[i+1]:
            decreasing = True

    return True

list1=[202,301,405,607,590,320,140,50]  
list2=[102,201,305,407,390,220,140,50]  
list3=[102,201,305,407,390,220,440,50]

print(hillnum(list1),
      
      hillnum(list2),
     
      hillnum(list3))
```

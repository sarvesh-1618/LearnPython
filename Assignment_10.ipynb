```python
# Q1) input=['hyd','bengaluru','mumbai','chennai']
#     output=['Hyd','Bengaluru','Mumbai','Chennai']

list1=['hyd','bengaluru','mumbai','chennai']
out=[]
for i in list1:
    out.append(i.capitalize())
out
```

```python
# Q2) input=['hyd','bengaluru','mumbai','chennai']
#     output=['HYD','BENGALURU','MUMBAI','CHENNAI']

list1=['hyd','bengaluru','mumbai','chennai']
out=[]
for i in list1:
    out.append(i.upper())
out
```

```python
# Q3) input=['hyd','bengaluru','mum#bai','chen#nai']
#     output=['mum#bai','chen#nai']
list1=['hyd','bengaluru','mum#bai','chen#nai']
out=[]
for i in list1:
    if '#' in i:
        out.append(i)
out
```

```python
# Q4) input=['hyd','bengaluru','mum#bai','chen#nai']
#     output=['hyd','bengaluru']

list1=['hyd','bengaluru','mum#bai','chen#nai']
out=[]
for i in list1:
    if '#'not in i:
        out.append(i)
out
```

```python
# Q5) input=['hyd','bengaluru','mumbai','chennai']
#     multiply all indexes by 10 
l1=['hyd','bengaluru','mumbai','chennai']
out=[]
for i in range(len(l1)):
    out.append(i*10)
out
```

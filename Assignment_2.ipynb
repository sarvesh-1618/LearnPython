```python
#1>>Total amount using simple intrest ??using format

p=eval(input("Enter principal : "))
t=eval(input("Enter time :"))
r=eval(input("Enter rate :"))

si=(p*t*r)/100
total=p+si

print("Simple intrest={}".format(si))   #format
print("Total amount={}".format(total))

print(f"simple intrest={si}")    #f-string
print(f"total amount={total}")
```

```python
#9>>18. Distance Between Two Points ??
x1=eval(input("Enter x1:"))
y1=eval(input("Enter y1:"))
x2=eval(input("Enter x2:"))
y2=eval(input("Enter y2:"))

d=((x2-x1)**2+(y2-y1)**2)*0.5

print("distance={}".format(d))
print(f"distance={d}")
```

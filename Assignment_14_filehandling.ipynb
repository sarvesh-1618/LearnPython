```python
# q3) extract lines which are starts with From,  append in a list
#  From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008/n
file=open(file_path)
d=[]
for i in file:
    i=i.strip()
    if i.startswith('From'):
        d.append(i)
print(d)
```

```python
# q4) from q3 you will see /n , Remove the /n at the end of the line append in a list
#    From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008
file=open(file_path)
d=[]
for i in file:
    i=i.rstrip()
    if i.startswith('From'):
        d.append(i)

print(d)
    
```

| Method     | Removes from Left | Removes from Right |
| ---------- | ----------------- | ------------------ |
| `strip()`  | ✅ Yes             | ✅ Yes              |
| `rstrip()` | ❌ No              | ✅ Yes              |

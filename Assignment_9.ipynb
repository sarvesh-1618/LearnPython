| Method           | Memory trick              | `True` when...             |
| ---------------- | ------------------------- | -------------------------- |
| `isalnum()`      | **AL**phabet + **NUM**ber | letters OR numbers         |
| `isalpha()`      | **ALPHA**                 | only letters               |
| `isascii()`      | **ASCII**                 | ASCII characters           |
| `isdecimal()`    | **DECIMAL**               | decimal characters `0-9`   |
| `isdigit()`      | **DIGIT**                 | digit characters           |
| `isidentifier()` | **IDENTIFIER**            | valid Python variable name |
| `islower()`      | **LOWER**                 | lowercase                  |
| `isnumeric()`    | **NUMERIC**               | numeric characters         |
| `isprintable()`  | **PRINT**                 | printable characters       |
| `isspace()`      | **SPACE**                 | spaces/whitespace          |
| `istitle()`      | **TITLE**                 | title-case text            |
| `isupper()`      | **UPPER**                 | uppercase                  |

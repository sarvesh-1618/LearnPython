```python
def palindrome(word):
    if len(word)<=1:
        return True 
        
    if word[0] != word[-1]:
        return False 
    
    return palindrome(word[1:-1])


word=input('Enter a word')
if palindrome(word):
    print('palindrome ')
else:
    print('Not palindrome ')
```

```python
def prime(*args):
    result=[]
    for i in args:
        if i > 1:
            for j in range(2,i):
                if i % j==0:
                    break 
            else:
                result.append(i)
    return result

prime(1,2,3,5,7,8)
    
```

```python
def count_character(text):
    uppercase=0
    lowercase=0
    digit=0

    for i in text:
        if i.isupper():
            uppercase = uppercase + 1
        elif i.islower():
            lowercase = lowercase +1
        elif i.isdigit():
            digit = digit + 1
        else:
            pass

    return uppercase,lowercase,digit

text=input('Enter a string :')
result = count_character(text)

            
print('uppercase=',result[0])
print('lowercase=',result[1])
print('digits=',result[2])
```

```python
def amstrong(num):
    original = num   
    power=len(str(num))
    total = 0

    while num  >0 :
        digit = num %  10
        total = total + (digit ** power) 
        num = num // 10

    if total == original :
       return True 
    else:
       return False
       
num=eval(input('Enter a number :'))

amstrong(num)
```

**global vs local**

```python
x=100
def show():
    y =200
    c= x+10
    print('c=',c)
    print('Global variable=',x)
    print('local variable=',y)

show()

print('Global variable =',x)
```

```python
var =20
def modify():
    global var
    var = var+ 100
    print('inside fun ,global var =',var)

print('before function call = ',var)

modify()

print('after function call =',var )
```

```python
a = 100
def observe():
    a =200
    print('inside function a =',a)
    
print('before function call a =',a)

observe()

print('after function call a = ',a)
```

```python
counter = 0
def fun():
    global counter
    counter += 1
    print('function called' ,counter,'times')
    
fun()
fun()
fun()
print('final counter=',counter)
```

```python
var = 100
def fun():                                     # x is accessible only inside the function.
    x=200
    print('local x=',x)
    print('global var=',var)
print('gloabl var =',var)
fun()
```

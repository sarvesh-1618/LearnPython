```python
#1. WAP to ask the user to take two numbers and print subtraction, multiplication, and division.
import  random 
a=random.randint(1,100)
b=random.randint(1,100)
print("a",a)
print("b",b)

print("substraction=",a-b)
print("multiplication=",a*b)
print("division=",round(a/b))
```

```python
#2. WAP to ask the user to take the radius and find the area of a circle.
import  math
import random

r=random.randint(1,20)

area=math.pi*r*r

print("Radious :",r)
print("area of the circole",round(area,3))
```

```python
#3. WAP to ask the user to take height and breadth and find the area of a triangle.
b=eval(input("Enter breadh :"))
h=eval(input("Enter height :"))

area=0.5*b*h

print("area of triangle",round(area,3))
```

```python
###### 5. WAP to ask the user to take the bill amount and tip amount and calculate the total bill.
bill=eval(input("Enter bill amount :"))
tip=eval(input("Enter tip amount :"))

total=bill+tip

print("total bill=",round(total,2))
```

```python
#6. WAP to ask the user to take the salary and percentage of tax and calculate the tax amount.
salary = eval(input("Enter salary :"))
tax_percentage=eval(input("Enter tax percentage:"))

tax=salary*tax_percentage  /100

print("tax amount",round(tax,2))
```

```python
# reverse number ??
# n=int(input('Enter a number :'))


def reverse(n):
    rev=0
    
    while n>0:
        digit = n % 10
        rev = rev*10 + digit
        n = n //10
    
    return rev

reverse(121)
reverse(123)


#Python tries to interpret 098 as a Python integer literal. In Python, decimal integer literals cannot have leading zeros, so it raises a SyntaxError. 
# eval(2+3)=5 ,int(2+3) =value error 
```

```python
# Try-Except Inside a Function

def safe_divide(a,b):
    try:
        return a/b
    except ZeroDivisionError:
        return 'cant divide by zero'
    except TypeError:
        return 'invalid input type'
    # except Exception as e:
    #     print(e)

print(safe_divide(10,2))
print(safe_divide(10, 0))
print(safe_divide(10,'hello'))
print(safe_divide('abc',2))
```

```python
# Reverse Words in a Sentence

def reverse_word(sentence):
    word = sentence.split()
    return ' '.join(word[::-1])

print(reverse_word('python is easy'))
print(reverse_word('i love coding '))
```

```python
# Percentage Calculator Using Functions

def calculate_percentage(marks):
    
    total=sum(marks)
    percentage=(total / len(marks) )

    return percentage


def assign_grade(percentage):
    if percentage >= 90:
        return 'A+'
    elif percentage >=80:
        return 'A'
    elif percentage >=70:
        return 'B'
    elif percentage >=60:
        return 'C'
    elif percentage >=50:
        return 'D'
    else:
        return 'fail'

marks=eval(input('Enter marks :'))


percentage=calculate_percentage(marks)

print('percentage :',percentage)

print(assign_grade(percentage))
```

```python
#Add Two Letters Using chr() and ord()

def add_letters(a,b):
    value1=ord(a)
    value2= ord(b)

    total =value1+value2

    return total ,chr(total)

print(add_letters('A','B'))
print(add_letters('X','Y'))
print(add_letters('a','z'))
```

```python
# Simple & Compound Interest
def interest(p,r,t):
    
    si=(p*t*r) /100

    amount=p*((1+r/100)**t)
    
    ci=amount - p  #Amount = Principal + Interest
    
   
    #return simple_interest,compound_interest
    print(f'simple interest={si:.2f},compound interest={ci:.2f}')

    
interest(1000,5,3)

interest(5000,7.5,2)

interest(20000,4.5,5)
    
#si,ci=interest(10000,5,3)
#print(f'simple interest={si},compound interest={ci}')
```

```python
# Armstrong Number Check (No Loops, No Recursion) ??

def is_armstrong(n):
    if 100<= n <=999:
        hundred =n // 100
        tens = (n % 100) // 10
        ones = n % 10

        total = hundred**3 + tens**3 + ones**3
        if total == n:
            print(f'True({hundred}^3 + {tens}^3 + {ones}^3 ={total})')
        else:
            print(f'False({hundred}^3 + {tens}^3 + {ones}^3 ={total})')

    elif 1000<=n<=9999:
        thousands= n //1000
        hundred =n % 1000 // 100
        tens = (n % 100) // 10
        ones = n % 10
    
        total = thousands**4 + hundred**4 + tens**4 + ones**4
        if total == n:
            print(f'True({thousands}^4+{hundred}^4 + {tens}^4 + {ones}^4 ={total})')
        else:
            print(f'False({thousands}^4 +{hundred}^4 + {tens}^4 + {ones}^4 ={total})')

        
is_armstrong(153)
is_armstrong(370)
is_armstrong(9474)
```

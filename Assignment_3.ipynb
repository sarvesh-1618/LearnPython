```python
#1. Ramesh’s basic salary is input through the keyboard.
#His dearness allowance is 40% of basic salary, and
#house rent allowance is 20% of basic salary.
#Write a program to calculate his gross salary.

basic_salary=eval(input("Enter basic salary :"))

DA=0.40*basic_salary
HRA=0.20*basic_salary

gross_salary=DA+HRA+basic_salary
print(gross_salary)
```

```python
#2. The distance between two cities (in km.) is input through the keyboard.
#Write a program to convert and
#print this distance in meters, feet, inches and centimeters.
km=eval(input("Enter distance in kilimeter : "))

meters=km * 1000
feet=km * 3280
inches=km * 39370
centimeter=km * 100000

print(meters)
print(feet)
print(inches)
print(centimeter)
```

```python
#4. Temperature of a city in Fahrenheit degrees is input through the keyboard.
#Write a program to convert this temperature into Centigrade degrees.

fahrenheit=eval(input("Enter fahrenheit value :"))

centigrade=(5/9)*(fahrenheit-32)

print(centigrade)
```

```python
#. In a town, the percentage of men is 52. The percentage of total literacy is 48. 
#If total percentage of literate men is 35 of the total population, 
#write a program to find the total number of illiterate men and women if the population of the town is 80,000.

population=80000

men=(52*population) / 100
women=population - men 

literate_people=(48*population )/100
literate_men=(35*population )/100

literate_women=literate_people - literate_men

illiterate_men= men - literate_men
illiterate_women= women - literate_women 

print(illiterate_men)
print(illiterate_women)
```

```python
#10. A cashier has currency notes of denominations 10, 50 and 100.
#If the amount to be withdrawn is input through the keyboard in hundreds,
#find the total number of currency notes of each denomination the cashier will have to give to the withdrawer.
```

```python
amount=eval(input("Enter the amount :"))

hundred = amount //100
remaining_money = amount - (hundred*100)
fifty = remaining_money // 50
remaining_money = remaining_money - (fifty*50)
tens = remaining_money // 10
print(f'100 X {hundred}')
print(f'50 X {fifty}')
print(f'10 X {tens}')
```
